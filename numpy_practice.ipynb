```python
import numpy as np

# Columns: [Driver_ID, Lap_Number, Sector_1_Time, Sector_2_Time, Sector_3_Time]
# Times are in seconds.
data = np.array([
    [1, 1, 28.5, 35.2, 24.1],
    [1, 2, 28.1, 34.8, 23.9],
    [1, 3, 27.9, 35.0, 24.0],
    [4, 1, 28.6, 35.4, 24.3],
    [4, 2, 28.2, 34.9, 24.0],
    [4, 3, 28.0, 34.7, 23.8]
])
```

```python
print(data)
```

```python
total_lap=np.array([data[:,2:]])
print(total_lap)
```

```python
total_lap_times=np.sum(total_lap[0],axis=1)
print(total_lap_times)
```

```python
fastest_lap=min(total_lap_times)
print(fastest_lap)
```

Satellite Image Thresholding


```python
import numpy as np
from skimage import data
import matplotlib.pyplot as plt


# Load a built-in grayscale image of a camera man as a 2D numpy array
image_array = data.camera()
```

```python
plt.imshow(image_array,cmap="gray");
```

```python
plt.show()
```

```python
print(image_array)
```

```python
min_pixel=np.min(image_array)
```

```python
print(min_pixel)
```

```python
max_pixel=np.max(image_array)
```

```python
print(max_pixel)
```

```python
mean_pixel=np.mean(image_array)
```

```python
print(mean_pixel)
```

```python
binary_mask=np.where(image_array>150,1,0)
print(binary_mask)
```

```python
target_pixels=np.sum(binary_mask)
print(target_pixels)
```

```python
total_pixels=image_array.size
print(total_pixels)
```

```python
target_percentage=(target_pixels/total_pixels) *100
```

```python
print(target_percentage)
```

```python
# 1D array of session times in seconds
session_times = np.array([120.5, 45.2, np.nan, 310.1, np.nan, 15.0, 88.5])
```

```python
nan_count=np.sum(np.isnan(session_times))
print(nan_count)
```

```python
nan_mean=np.nanmean(session_times)
print(nan_mean)
```

```python
new_session_times=np.where(np.isnan(session_times),nan_mean,session_times)
print(new_session_times)
```

```python
# Raw 1D data stream from the client (x1, y1, z1, x2, y2, z2, x3, y3, z3)
raw_stream = np.array([1.0, 2.0, 1.0, 0.0, 1.0, 1.0, 3.0, 0.0, 1.0])

# 3x3 Transformation Matrix (Scaling by a factor of 2)
scale_matrix = np.array([
    [2.0, 0.0, 0.0],
    [0.0, 2.0, 0.0],
    [0.0, 0.0, 2.0]
])
```

```python
vertices=raw_stream.reshape(3,3)
```

```python
print(vertices)
```

```python
transformed_vertices=scale_matrix*vertices
print(transformed_vertices)
```

```python
# Rows: Portfolio 1, Portfolio 2, Portfolio 3
# Columns: Shares of [AAPL, GOOGL, AMZN]
portfolios = np.array([
    [10, 5, 20],
    [0, 50, 10],
    [100, 0, 0]
])

# Current prices: [AAPL, GOOGL, AMZN]
prices = np.array([170.0, 140.0, 180.0])
```

```python
total=portfolios @ prices
print(total)
```

```python
total_portfolio=np.dot(portfolios,prices)
```
